## Task 2: Color Space Conversion & Histogram

```python
import cv2
import matplotlib.pyplot as plt
```

```python
# Load image
image = cv2.imread("/content/Beautiful girl.jpg")
```

```python
# Convert to different color spaces
gray_image = cv2.cvtColor(image, cv2.COLOR_BGR2GRAY)
hsv_image = cv2.cvtColor(image, cv2.COLOR_BGR2HSV)
lab_image = cv2.cvtColor(image, cv2.COLOR_BGR2LAB)
```

```python
# Save converted images
cv2.imwrite("photo_grayscale.jpg", gray_image)
cv2.imwrite("photo_hsv.jpg", hsv_image)
cv2.imwrite("photo_lab.jpg", lab_image)
```

```python
# Display images
from google.colab.patches import cv2_imshow

cv2_imshow(gray_image)
cv2_imshow(hsv_image)
cv2_imshow(lab_image)
```

```python
# Plot histogram for grayscale image
plt.hist(gray_image.ravel(), bins=256, range=[0, 256])
plt.title("Grayscale Histogram")
plt.xlabel("Pixel Intensity")
plt.ylabel("Frequency")
plt.show()

cv2.waitKey(0)
cv2.destroyAllWindows()
```
